```python
X=np.array([[1,0],[2,0],[3,0],[-1,0],[-2,0],[-3,0],[2,-1],[-2,1]])
# print(X)
X_centred=X-np.mean(X,axis=0)
# print(X_centred) # in our case X and X_centred are same
X_T=np.transpose(X_centred)
# print(X_T)
C=(1/8)*np.matmul(X_T,X_centred) # covariance matrix
print(C)
```

```python
eigen_values,eigen_vectors=np.linalg.eig(C)
print(eigen_values)
print(eigen_vectors)
print(eigen_vectors[0])
```
